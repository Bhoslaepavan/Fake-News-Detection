#### Data Cleaning 

```python
import nltk
nltk.download("stopwords")
```

```python
from nltk.corpus import stopwords
```

```python
spotpwords=list(stopwords.words("english"))
```

```python
spotpwords
```

```python
def convert_lower(text):
    text=text.lower()
    return text
```

```python
df_shuffle["new_text"]=df_shuffle["new_text"].apply(convert_lower)
```

```python
df_shuffle.head(5)
```

```python
import re
import string
```

```python
exclude=string.punctuation
```

```python
def punctuation(text):
    for char in text:
        if char in exclude:
            text=text.replace(char,"")
        return text
```

```python
df_shuffle["new_text"]=df_shuffle["new_text"].apply(punctuation)
```

```python
def remove_stopwords(text):
    lst=[]
    for i in text.split():
        if i not in spotpwords:
            lst.append(i)
    return " ".join(lst)
  
```

```python
df_shuffle["new_text"]=df_shuffle["new_text"].apply(remove_stopwords)
```

```python
df_shuffle
```

```python
def char_count(text):
    count=0
    for ch in text:
        count+=1
    return count
```

```python
df_shuffle["char_count"]=df_shuffle["new_text"].apply(char_count)
```

```python
df_shuffle.head(5)
```

```python
df_shuffle.head(5)
```

```python
def word_count(text):
    return len(text.split())
```

```python
df_shuffle["word_count"]=df_shuffle["new_text"].apply(word_count)
```

```python
# Feature enginerring usung the tfidf
```

```python
from sklearn.model_selection import train_test_split
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
x=df_shuffle['new_text']
y=df_shuffle['label']
```

```python
# Split data (80% train, 20% test)
x_train,x_test,y_train,y_test = train_test_split(x,y,test_size=0.3,random_state=42)
```

```python
#i ising the tfidf 
```

```python
tfidf=TfidfVectorizer(max_features=5000)
```

```python
# train data and the testiing data trasforming 
```

```python
x_train_tfidf = tfidf.fit_transform(x_train)
x_test_tfidf  =tfidf.transform(x_test)
```

```python
from sklearn.linear_model import LogisticRegression
```

```python
from sklearn.metrics import accuracy_score, classification_report
```

```python
#model bulid ani train dataset
```

```python
model = LogisticRegression() 
```

```python
model.fit(x_train_tfidf,y_train)
```

```python
#test in test data 
```

```python
y_pred = model.predict(x_test_tfidf)
```

```python
accuracy_score(y_test, y_pred)
```

```python
classification_report(y_test, y_pred)
```

```python
# Training set var prediction kara  # model not ovefit beacude the train and testing acuracy not big diff
y_train_pred = model.predict(x_train_tfidf)

print("Training Accuracy:", accuracy_score(y_train, y_train_pred))
print("Testing Accuracy:", accuracy_score(y_test, y_pred))
```
